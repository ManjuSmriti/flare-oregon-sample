# Open and check FLARE Outlook v11

Choose one of the 19 daily forecasts. Every file contains the full native grid and its latitude/longitude. No original Zarr stores are required. Install the package requirements before running.

```python
%matplotlib inline
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'manifest.json').exists():
    ROOT = ROOT.parent
assert (ROOT / 'manifest.json').exists(), 'Run from the package root or notebooks folder.'
sys.path.insert(0, str(ROOT / 'examples'))
import outlook
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
index = outlook.catalog()
index[['date', 'issue', 'valid_start', 'valid_end', 'bytes']]
```

```python
DATE = '2026-07-03'  # Change to any date through 2026-07-21.
assert outlook.verify_file(DATE)
ds = outlook.open_day(DATE)
assert ds.sizes == {'y': 5792, 'x': 6672}
assert set(ds.data_vars) == {'outlook_index', 'category'}
assert ds.category.dtype == np.dtype('uint8')
ds
```

Latitude and longitude are **2D auxiliary coordinates**. Integer class 0 is retained as unavailable. Forecast times are attributes; the processing date is separate from the July issue date.

```python
print('Issue:', ds.attrs['forecast_issue_time'])
print('Valid:', ds.attrs['time_coverage_start'], 'to', ds.attrs['time_coverage_end'])
print('Category:', ds.category.attrs['flag_meanings'])
print('Latitude range:', float(ds.lat.min()), float(ds.lat.max()))
print('Longitude range:', float(ds.lon.min()), float(ds.lon.max()))
ds.close()
```
